### BoW - 
- describes the occurence of words within a document completely ignoring the order
- one of the simplest text representations
- counts how many times a word appears in a documents

```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

corpus = [
    "Forests are green and beautiful.",
    "Green trees make a forest beautiful"
]

vectorizer = CountVectorizer()
bow_matrix = vectorizer.fit_transform(corpus) # Sparse Row sparse matrix of dtype 'int64'

bow_matrix.toarray() # convert to array

pd.DataFrame(bow_matrix.toarray(), columns = vectorizer.get_feature_names_out())
```

- calculates frequencies of each of the words
- more repetition of a word, means more important it is

### TF-IDF - 
- evaluates importance of word, relative to the collection of documents (corpus)
- relative to whole of the text, how imporatant is a word

- It rewards the words that appear a lot in a specific `document` across the `corpus` ---> Term Frequency
- ![](https://media.geeksforgeeks.org/wp-content/uploads/20250207113146020518/The-TF-Formula.webp)
- it heavily penalizes words that appear in every document ---> Inverse Document Frequency
- penalizes more frequency
- ![](https://media.geeksforgeeks.org/wp-content/uploads/20250207113218469447/IDF-Formula.webp)
- 
- example:
    - 1000 articles
    - `the` appears everywhere in all the corpus ---> tfidf score = 0
    - word `jedi` appears 40 times in a specific article
    - tfidf will give this term a massive score ---> means this term is important for that document
    - `jedi` ---> article is related to starwars


![](https://media.geeksforgeeks.org/wp-content/uploads/20250813164205496346/idf_formula.webp)
- 

```python
corpus = [
    "Forests are green and beautiful.",
    "Green trees make a forest beautiful"
]

from sklearn.feature_extraction.text import TfidfVectorizer

tf_idf = TfidfVectorizer()

tfidf_matrix = tf_idf.fit_transform(corpus)

tfidf_matrix.toarray() # tf-idf scores

pd.DataFrame(tfidf_matrix.toarray(), columns= tf_idf.get_feature_names_out())

idfs = tf_idf.idf_
words = tf_idf.get_feature_names_out()

pd.DataFrame(idfs.T, words)
```

```python
tf_idf.vocabulary_
```

```python
import spacy
nlp = spacy.load("en_core_web_sm")

word_1 = nlp("king")
word_2 = nlp("queen")
word_3 = nlp("man")
word_4 = nlp("woman")

print("king queen: ", word_1.similarity(word_2))
print("man woman: ", word_3.similarity(word_4))
print("man queen: ", word_3.similarity(word_2))
print("man king: ", word_3.similarity(word_1))
print("woman queen: ", word_2.similarity(word_4))
```

### Character Encoding
- text into binary
- We have international standards for this
- 
- ASCII - 
    - 7 bit for character encoding
    - (A-Z, a-z, 0-9), and some basic punctuations as well
    - did not support emojis, special characters, non english, 
- UTF-8 - 
    - uses variable width 
    - uses 1 byte for standard english ---> makes it compatible with ASCII
    - 
- UTF - 16 ---> uses 2 to 4 bytes, UTF-32 ---> fixed 4 bytes

```python
ord('A') ### ASCII code

letters = ['A', 'B', 'C', 'a', 'b', 'c', '!', '1','2','3']
[ord(w) for w in letters] # ASCII codes
```

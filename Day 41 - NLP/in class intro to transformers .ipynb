### Chatbot - Local Text Generation

```python
from transformers import pipeline
generator = pipeline("text-generation", model = "gpt2")
```

```python
text = "who was muamar gaddafi?"

result = generator(
    text,
    max_new_tokens = 100,
    temperature = 0.2, # risk setting ---> how safe/boring
    # 0.5 or 0.7 is balanced but 1.7 or above is wild/unpredicted
)
print(result[0]['generated_text'])
```

## Gemini Chatbot

```python
# pip install google-genai python-dotenv
```

```python
import os
from dotenv import load_dotenv

load_dotenv()

api_key = os.getenv("google_api_key")
```

```python
from google import genai

client = genai.Client(api_key= api_key)

chat = client.chats.create(model = "gemini-2.5-flash-lite")

# "gemini-2.5-flash" ~ 10 -20B - 1M - 15 RPM / 1,000 RPD - 
# "gemini-2.5-flash-lite" ~ 5-8B - 65K - 30 RPM / 2,000 RPD 
# "gemma-3n-e2b-it" - 2B - 32K
# gemma-3-27b-it - 27B - 128K
```

```python
prompt = "who is muammar gaddafi?"
prompt = "who is Mohammad Idrees?"

response = chat.send_message(prompt)

print(response.text)
```

```python
while True:
    user_input = input("Prompt: ")

    if user_input.lower() in ["quit", "exit", "bye"]:
        break
    else:
        response = chat.send_message(user_input)

    print(f'Gemini:=========\n {response.text}\n')
```

```python
import ollama


prompt = "who was gaddafi"
prompt = "your last training date"

response = ollama.chat(
    model = 'llama3.2',
    # stream= True,
    options= {
        'temperature': 0.1,
        'num_ctx': 4096,
    },
    messages=[
        {
            'role':'user',
            'content': prompt
        }
    ]
)
```
